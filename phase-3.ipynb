```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structure : [Restaurtent_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],# Paradise Biryani
    [2, 120000, 140000, 160000, 190000],# Bejing Bites
    [3, 200000, 230000, 260000, 300000],# Pizza Hub
    [4, 180000, 210000, 240000, 270000],# Burger Point    
    [5, 160000, 185000, 205000, 230000] # Chai Point
])
print("======Zomato sales analasis======")

print("Sales data shape is:",sales_data.shape)

print("the first three restaurent is:\n",sales_data[0:3])
```

```python
# Total sales per year 
print("The total sales per year is:\n",sales_data[:,4:]) #here : = rows:column and then , and again rows:column 4: = range 

# Total sales in 2021 and 2024
print("total sales in 2021 and 2022 is \n",sales_data[:,1:3])

# Sum of 1 st column
print(np.sum(sales_data[0,1:6]))

# Total sales per year
print(np.sum(sales_data[:1,]))
```

```python
# Total sales of all restaurants for each year
print(np.sum(sales_data[:,1:],axis=1))

# Total sales of each restaurant across all years 
print (np.sum(sales_data[:,1:],axis=0))

# Minimum sales per resraurents
print(np.min(sales_data[:,1:],axis=1))

# Maximum sales per years
print(np.max(sales_data[:,1:],axis=0))

# Average sales per resturents
print(np.average(sales_data[:,1:],axis=1))

cumsum = np.cumsum(sales_data[:,1:],axis=1)
print(cumsum)

plt.figure(figsize =(8,6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average cumelative sum across all restaurent")
plt.xlabel("years")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```
