### creating arrays from list

### list vs numpy array

```python
py_list = [1,2,3,4]
print("\n ptyhon list multiplication",py_list*2)

import numpy as np
np_array = np.array([1,3,4])#element wise multiplication
print("python multiplaction :", np_array*2)

import time
start = time.time()
py_list = [i*2 for i in range(100000)]
print("\n List opreation time: ", time.time() - start)

start = time.time()
np.array = np.arange(100000)*2
print("\n Numpy opreation time: ", time.time() - start)
```

### creating array from scratch

```python
import numpy as np
zeros = np.zeros((3,4)) # it will print 3x4 matrices only 0 will print
print("zeros in array:\n",zeros)

ones = np.ones((3,5))# it will only print 1 in the 3x5 matrix    
print("ones in array:\n",ones)

full = np.full((2,3), 7)
print("full array which contain only 7 :\n", full)

random = np.random.random((2,2))
print("the random in array:\n", random)

sequence = np.arange(0,10)
print("the sequence array is:\n",sequence)
```

### vector , matrix and tensor 

```python
vector = np.array([1,2,3])
print("the vector is:",vector)

matrix = np.array([[1,2,3],
                   [4,5,6]])
print("matrix is:",matrix)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]])
print("tensor is:\n", tensor)
```

### Array Properties

```python
array = np.array([[2,3,4],
                  [4,5,6]])
print("shape:",array.shape)
print("size:",array.size)
print("data type:",array.dtype)
print("dimenson:",array.ndim)
```

### Array Reshaping

```python
import numpy as np
arr = np.arange(12)
print("Original array:",arr)

reshape = arr.reshape((3,4))
print("reshaped array:",reshape)

# flattened copy the original array
flattened = reshape.flatten()
print("the falttened array is:",flattened)

#reveled view instead of copy
reveled = flattened.ravel()
print("the reveled array is:",reveled)

transposed = reshape.T
print("the reshaaped array is:",transposed)
```
